## 1. Setup and Configuration

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings

warnings.filterwarnings('ignore')

# Set visualization styles
sns.set_style('whitegrid')
sns.set_palette('husl')
plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 12

print("✓ Libraries loaded successfully")
print("✓ Visualization settings configured")
```

## 2. Data Preparation

```python
# Create sample dataset for visualization
np.random.seed(42)

# Generate monthly data for 2 years
months = pd.date_range(start='2022-01-01', end='2023-12-31', freq='MS')

data = {
    'Month': months,
    'Revenue': np.cumsum(np.random.randn(24) * 5000 + 50000),
    'Expenses': np.cumsum(np.random.randn(24) * 3000 + 30000),
    'Customers': np.cumsum(np.random.randint(-50, 150, 24) + 100),
    'Product_A_Sales': np.random.randint(1000, 5000, 24),
    'Product_B_Sales': np.random.randint(800, 4000, 24),
    'Product_C_Sales': np.random.randint(600, 3500, 24),
}

df = pd.DataFrame(data)
df['Profit'] = df['Revenue'] - df['Expenses']
df['Profit_Margin'] = (df['Profit'] / df['Revenue'] * 100).round(2)

print("Dataset created successfully!")
print(f"Shape: {df.shape}")
df.head()
```

## 3. Line Charts - Trends Over Time

```python
# Multi-line chart showing revenue and expenses over time
fig, ax = plt.subplots(figsize=(16, 8))

ax.plot(df['Month'], df['Revenue'], marker='o', linewidth=2.5, 
        label='Revenue', color='#2ECC71', markersize=6)
ax.plot(df['Month'], df['Expenses'], marker='s', linewidth=2.5, 
        label='Expenses', color='#E74C3C', markersize=6)
ax.plot(df['Month'], df['Profit'], marker='^', linewidth=2.5, 
        label='Profit', color='#3498DB', markersize=6)

ax.set_title('Financial Performance Over Time (2022-2023)', 
             fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Month', fontsize=13, fontweight='bold')
ax.set_ylabel('Amount ($)', fontsize=13, fontweight='bold')
ax.legend(loc='upper left', fontsize=12, framealpha=0.9)
ax.grid(True, alpha=0.3, linestyle='--')

# Add a shaded region for profitable months
profitable_months = df['Profit'] > 0
ax.axhline(y=0, color='black', linestyle='-', linewidth=0.8, alpha=0.5)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\n📊 Insight: Revenue shows steady growth with seasonal fluctuations")
print("💡 Profit margin remains healthy throughout the period")
```

## 4. Bar Charts - Categorical Comparisons

```python
# Product sales comparison
product_totals = {
    'Product A': df['Product_A_Sales'].sum(),
    'Product B': df['Product_B_Sales'].sum(),
    'Product C': df['Product_C_Sales'].sum()
}

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Vertical bar chart
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']
axes[0].bar(product_totals.keys(), product_totals.values(), color=colors, edgecolor='black', linewidth=1.5)
axes[0].set_title('Total Sales by Product (2022-2023)', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Total Units Sold', fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.3, axis='y')

# Add value labels on bars
for i, (product, value) in enumerate(product_totals.items()):
    axes[0].text(i, value, f'{value:,}', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Horizontal bar chart - Average monthly sales
avg_sales = {
    'Product A': df['Product_A_Sales'].mean(),
    'Product B': df['Product_B_Sales'].mean(),
    'Product C': df['Product_C_Sales'].mean()
}

axes[1].barh(list(avg_sales.keys()), list(avg_sales.values()), color=colors, edgecolor='black', linewidth=1.5)
axes[1].set_title('Average Monthly Sales by Product', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Average Units Sold per Month', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3, axis='x')

# Add value labels
for i, (product, value) in enumerate(avg_sales.items()):
    axes[1].text(value, i, f'{value:,.0f}', ha='left', va='center', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n📊 Insight: Product A is the top performer in both total and average sales")
print("💡 Consider investigating why Product C underperforms")
```

## 5. Scatter Plots - Relationships

```python
# Relationship between revenue and customers
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Scatter plot with trend line
axes[0].scatter(df['Customers'], df['Revenue'], s=100, alpha=0.6, 
                c=df['Profit_Margin'], cmap='viridis', edgecolors='black', linewidth=1)

# Add trend line
z = np.polyfit(df['Customers'], df['Revenue'], 1)
p = np.poly1d(z)
axes[0].plot(df['Customers'], p(df['Customers']), "r--", linewidth=2, label='Trend Line')

axes[0].set_title('Revenue vs. Customer Count', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Number of Customers', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Revenue ($)', fontsize=12, fontweight='bold')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

cbar = plt.colorbar(axes[0].collections[0], ax=axes[0])
cbar.set_label('Profit Margin (%)', fontsize=11)

# Scatter plot: Profit Margin vs Revenue
axes[1].scatter(df['Revenue'], df['Profit_Margin'], s=100, alpha=0.6,
                c=range(len(df)), cmap='coolwarm', edgecolors='black', linewidth=1)
axes[1].set_title('Profit Margin vs. Revenue', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Revenue ($)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Profit Margin (%)', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Calculate correlation
correlation = df['Customers'].corr(df['Revenue'])
print(f"\n📊 Correlation between Customers and Revenue: {correlation:.3f}")
print("💡 Strong positive relationship indicates customer acquisition drives revenue")
```

## 6. Heatmap - Correlation Matrix

```python
# Create correlation matrix
correlation_data = df[['Revenue', 'Expenses', 'Profit', 'Customers', 
                        'Product_A_Sales', 'Product_B_Sales', 'Product_C_Sales']].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_data, annot=True, fmt='.2f', cmap='RdYlGn', 
            center=0, square=True, linewidths=2, cbar_kws={"shrink": 0.8},
            vmin=-1, vmax=1)

plt.title('Correlation Matrix - Business Metrics', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

print("\n📊 Insight: Heatmap reveals strong correlations between key business metrics")
print("💡 Revenue and expenses show expected positive correlation")
print("💡 Product sales show varying degrees of correlation with overall revenue")
```

## 7. Box Plots - Distribution Analysis

```python
# Prepare data for box plot
product_data = pd.DataFrame({
    'Product': ['A'] * len(df) + ['B'] * len(df) + ['C'] * len(df),
    'Sales': list(df['Product_A_Sales']) + list(df['Product_B_Sales']) + list(df['Product_C_Sales'])
})

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Box plot
sns.boxplot(data=product_data, x='Product', y='Sales', palette='Set2', ax=axes[0])
axes[0].set_title('Sales Distribution by Product', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Sales (Units)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Product', fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.3, axis='y')

# Violin plot for more detail
sns.violinplot(data=product_data, x='Product', y='Sales', palette='muted', ax=axes[1])
axes[1].set_title('Sales Distribution (Violin Plot)', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Sales (Units)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Product', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print("\n📊 Insight: Box plots reveal the distribution and variability of sales for each product")
print("💡 Product A shows higher median sales and wider distribution")
print("💡 Outliers visible in the data indicate occasional exceptional sales periods")
```

## 8. Stacked Area Chart - Composition Over Time

```python
# Stacked area chart for product sales contribution
plt.figure(figsize=(16, 8))

plt.stackplot(df['Month'], 
              df['Product_A_Sales'], 
              df['Product_B_Sales'], 
              df['Product_C_Sales'],
              labels=['Product A', 'Product B', 'Product C'],
              colors=['#FF6B6B', '#4ECDC4', '#45B7D1'],
              alpha=0.8)

plt.title('Product Sales Composition Over Time', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Month', fontsize=13, fontweight='bold')
plt.ylabel('Total Sales (Units)', fontsize=13, fontweight='bold')
plt.legend(loc='upper left', fontsize=12)
plt.grid(True, alpha=0.3, linestyle='--')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\n📊 Insight: Stacked area chart shows how each product contributes to total sales")
print("💡 Product A consistently maintains the largest share")
print("💡 Overall sales volume trend is visible through total area")
```

## 9. Dual-Axis Chart - Multiple Metrics

```python
# Dual-axis chart: Revenue and Profit Margin
fig, ax1 = plt.subplots(figsize=(16, 8))

# Primary axis - Revenue
color1 = '#2E86AB'
ax1.plot(df['Month'], df['Revenue'], color=color1, linewidth=3, marker='o', markersize=7, label='Revenue')
ax1.set_xlabel('Month', fontsize=13, fontweight='bold')
ax1.set_ylabel('Revenue ($)', fontsize=13, fontweight='bold', color=color1)
ax1.tick_params(axis='y', labelcolor=color1)
ax1.grid(True, alpha=0.3)

# Secondary axis - Profit Margin
ax2 = ax1.twinx()
color2 = '#E63946'
ax2.plot(df['Month'], df['Profit_Margin'], color=color2, linewidth=3, 
         marker='s', markersize=7, label='Profit Margin', linestyle='--')
ax2.set_ylabel('Profit Margin (%)', fontsize=13, fontweight='bold', color=color2)
ax2.tick_params(axis='y', labelcolor=color2)

# Title and legend
plt.title('Revenue and Profit Margin Trends', fontsize=16, fontweight='bold', pad=20)
fig.legend(loc='upper left', bbox_to_anchor=(0.12, 0.88), fontsize=12)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\n📊 Insight: Dual-axis chart allows comparison of metrics with different scales")
print("💡 Revenue growth doesn't always correlate with profit margin improvement")
print("💡 This highlights the importance of monitoring multiple KPIs")
```

## 10. Dashboard Summary

```python
# Create a comprehensive dashboard
fig = plt.figure(figsize=(18, 12))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# 1. Revenue Trend (top left, spans 2 columns)
ax1 = fig.add_subplot(gs[0, :2])
ax1.plot(df['Month'], df['Revenue'], linewidth=2.5, color='#2ECC71', marker='o')
ax1.set_title('Revenue Trend', fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.tick_params(axis='x', rotation=45)

# 2. Profit Margin (top right)
ax2 = fig.add_subplot(gs[0, 2])
ax2.bar(range(len(df)), df['Profit_Margin'], color='#3498DB', alpha=0.7)
ax2.set_title('Profit Margin Distribution', fontweight='bold')
ax2.set_ylabel('%')
ax2.grid(True, alpha=0.3, axis='y')

# 3. Product Sales (middle, spans all columns)
ax3 = fig.add_subplot(gs[1, :])
ax3.stackplot(df['Month'], df['Product_A_Sales'], df['Product_B_Sales'], df['Product_C_Sales'],
              labels=['Product A', 'Product B', 'Product C'],
              colors=['#FF6B6B', '#4ECDC4', '#45B7D1'], alpha=0.8)
ax3.set_title('Product Sales Mix', fontweight='bold')
ax3.legend(loc='upper left')
ax3.grid(True, alpha=0.3)
ax3.tick_params(axis='x', rotation=45)

# 4. Customer Growth (bottom left)
ax4 = fig.add_subplot(gs[2, 0])
ax4.fill_between(df['Month'], df['Customers'], alpha=0.5, color='#9B59B6')
ax4.plot(df['Month'], df['Customers'], linewidth=2, color='#8E44AD', marker='o')
ax4.set_title('Customer Growth', fontweight='bold')
ax4.grid(True, alpha=0.3)
ax4.tick_params(axis='x', rotation=45)

# 5. Total Product Sales (bottom middle)
ax5 = fig.add_subplot(gs[2, 1])
product_totals_list = [df['Product_A_Sales'].sum(), df['Product_B_Sales'].sum(), df['Product_C_Sales'].sum()]
ax5.pie(product_totals_list, labels=['Product A', 'Product B', 'Product C'],
        autopct='%1.1f%%', colors=['#FF6B6B', '#4ECDC4', '#45B7D1'], startangle=90)
ax5.set_title('Total Sales Share', fontweight='bold')

# 6. Key Metrics (bottom right)
ax6 = fig.add_subplot(gs[2, 2])
ax6.axis('off')
metrics_text = f"""
KEY METRICS

Total Revenue:
${df['Revenue'].sum():,.0f}

Avg Profit Margin:
{df['Profit_Margin'].mean():.1f}%

Total Customers:
{df['Customers'].iloc[-1]:,}

Best Month:
{df.loc[df['Revenue'].idxmax(), 'Month'].strftime('%b %Y')}
"""
ax6.text(0.1, 0.5, metrics_text, fontsize=11, verticalalignment='center',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))

fig.suptitle('Business Performance Dashboard', fontsize=18, fontweight='bold', y=0.98)
plt.show()

print("\n" + "="*60)
print("📊 COMPREHENSIVE DASHBOARD CREATED")
print("="*60)
print("\nThis dashboard demonstrates:")
print("✓ Multiple visualization types in one view")
print("✓ Consistent color scheme and styling")
print("✓ Clear hierarchy of information")
print("✓ Professional layout and composition")
print("✓ Key metrics highlighted for quick insights")
```
